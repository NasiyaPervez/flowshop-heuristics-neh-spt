# NEH Heuristic Implementation

    The NEH (Nawaz–Enscore–Ham) heuristic minimizes makespan in flowshop scheduling. The NEH heuristic prioritizes jobs with longer processing times and uses an iterative insertion process to build a schedule that minimizes the makespan or other relevant objectives. 





Sort jobs in descending order of total processing time:

Order: Job 1 (127) → Job 5 (114) → Job 2 (111) → Job 3 (98) → Job 4 (62).

Iterative insertion:

Start with the first two jobs (1, 5). Test sequences [1] and [1], compute makespan for both, and keep the better one.

Insert Job 2 into all possible positions of the current sequence and retain the best partial sequence.

Repeat for Jobs 3 and 4.

```python
# NEH heuristic for your jobs_original instance.
    
import numpy as np

def calculate_makespan(sequence, jobs):
    """
    Calculate the makespan for a given sequence of jobs in a flowshop.
    jobs: dict of {job_id: [processing_times_on_machines]}
    sequence: list of job_ids in the order to be scheduled
    """
    n_jobs = len(sequence)
    n_machines = len(jobs[sequence[0]])
    # Initialize completion time table
    completion = np.zeros((n_jobs, n_machines), dtype=int)
    
    for i, job_id in enumerate(sequence):
        for m in range(n_machines):
            if i == 0 and m == 0:
                # First job, first machine
                completion[i, m] = jobs[job_id][m]
            elif i == 0:
                # First job, rest of machines
                completion[i, m] = completion[i, m-1] + jobs[job_id][m]
            elif m == 0:
                # First machine, rest of jobs
                completion[i, m] = completion[i-1, m] + jobs[job_id][m]
            else:
                # General case
                completion[i, m] = max(completion[i-1, m], completion[i, m-1]) + jobs[job_id][m]
    return completion[-1, -1]

def neh_heuristic(jobs):
    """
    NEH heuristic for permutation flowshop scheduling.
    jobs: dict of {job_id: [processing_times_on_machines]}
    Returns: best sequence (list of job_ids), best makespan (int)
    """
    # Step 1: Calculate total processing time for each job
    job_totals = {job_id: sum(times) for job_id, times in jobs.items()}
    # Step 2: Sort jobs in descending order of total processing time
    sorted_jobs = sorted(job_totals, key=job_totals.get, reverse=True)
    # Step 3: Build up the sequence iteratively
    sequence = []
    for job in sorted_jobs:
        best_seq = None
        best_makespan = None
        # Try inserting this job in every possible position
        for i in range(len(sequence)+1):
            trial_seq = sequence[:i] + [job] + sequence[i:]
            makespan = calculate_makespan(trial_seq, jobs)
            if best_makespan is None or makespan < best_makespan:
                best_makespan = makespan
                best_seq = trial_seq
        sequence = best_seq
    return sequence, best_makespan

# Example usage:
jobs_original = {
    1: [31, 41, 25, 30],
    2: [19, 55, 3, 34],
    3: [23, 42, 27, 6],
    4: [13, 22, 14, 13],
    5: [33, 5, 57, 19]
}

best_seq, best_makespan = neh_heuristic(jobs_original)
print("NEH Sequence (jobs_original):", best_seq)
print("NEH Makespan (jobs_original):", best_makespan)
```

# SPT Algorithm Implementation
The Shortest Processing Time (SPT) rule schedules jobs in ascending order of their total processing time.

Steps for jobs_original:
Sort jobs by total processing time (ascending):

Order: Job 4 (62) → Job 3 (98) → Job 2 (111) → Job 5 (114) → Job 1 (127).



```python
# How it works (jobs_original - SPT)
#calculate_makespan: Computes the total completion time (makespan) for a given sequence.

# spt_heuristic: Implements the SPT rule by sorting jobs in ascending order of their total processing time and then calculates the makespan for this sequence.

# Example usage: Runs the SPT heuristic for jobs_original instance and prints the sequence and makespan.

import numpy as np

def calculate_makespan(sequence, jobs):
    """
    Calculate the makespan for a given sequence of jobs in a flowshop.
    jobs: dict of {job_id: [processing_times_on_machines]}
    sequence: list of job_ids in the order to be scheduled
    """
    n_jobs = len(sequence)
    n_machines = len(jobs[sequence[0]])
    completion = np.zeros((n_jobs, n_machines), dtype=int)
    
    for i, job_id in enumerate(sequence):
        for m in range(n_machines):
            if i == 0 and m == 0:
                completion[i, m] = jobs[job_id][m]
            elif i == 0:
                completion[i, m] = completion[i, m-1] + jobs[job_id][m]
            elif m == 0:
                completion[i, m] = completion[i-1, m] + jobs[job_id][m]
            else:
                completion[i, m] = max(completion[i-1, m], completion[i, m-1]) + jobs[job_id][m]
    return completion[-1, -1]

def spt_heuristic(jobs):
    """
    SPT heuristic for permutation flowshop scheduling.
    jobs: dict of {job_id: [processing_times_on_machines]}
    Returns: SPT sequence (list of job_ids), makespan (int)
    """
    # Sort jobs by total processing time (ascending)
    sorted_jobs = sorted(jobs, key=lambda job_id: sum(jobs[job_id]))
    makespan = calculate_makespan(sorted_jobs, jobs)
    return sorted_jobs, makespan

# Example usage:
jobs_original = {
    1: [31, 41, 25, 30],
    2: [19, 55, 3, 34],
    3: [23, 42, 27, 6],
    4: [13, 22, 14, 13],
    5: [33, 5, 57, 19]
}

spt_seq, spt_makespan = spt_heuristic(jobs_original)
print("SPT Sequence (jobs_original):", spt_seq)
print("SPT Makespan (jobs_original):", spt_makespan)
```

# Interpretation of the Gantt Chart for Job Scheduling

A Gantt chart is a visual scheduling tool that displays jobs (or tasks) as horizontal bars along a timeline. In flowshop scheduling, each bar represents a job being processed on a particular machine, with its position and length indicating start time and duration.

## Job Scheduling Gantt Chart

- **Horizontal Axis (Time):**  
  Shows the progression of time.

- **Vertical Axis (Machines):**  
  Lists the machines (e.g., Machine 1, Machine 2, etc.).

- **Bars (Jobs):**  
  Each colored bar represents a job on a machine. The start of the bar is when the job begins, and the end is when it finishes. The bar’s length shows the processing time.

- **Job Order and Overlap:**  
  The order of bars shows the job sequence. Gaps indicate idle machine time; overlapping bars on different machines show concurrent processing.

## What the Chart Reveals

- **Job Sequences:**  
  Comparing NEH and SPT charts shows how job order affects efficiency.

- **Idle Times and Bottlenecks:**  
  Gaps in bars are idle times. Fewer gaps mean a more efficient schedule.

- **Makespan:**  
  The rightmost end of the last bar on the last machine is the makespan (total completion time). Shorter is better.

## Example Interpretation

- **NEH Gantt Chart:**  
  Bars are packed more tightly, with less idle time and a shorter makespan.

- **SPT Gantt Chart:**  
  More gaps and a longer makespan.

## Why This Matters

Reading the Gantt chart helps you:
- Visually compare scheduling strategies (NEH vs. SPT)
- Identify which approach uses machines more efficiently
- Spot bottlenecks and optimize further

# Mixed-Integer Programming (MIP) formulation

```python
from gurobipy import Model, GRB
import itertools

# --- Job instance: jobs_original ---
jobs = {
    1: [31, 41, 25, 30],
    2: [19, 55, 3, 34],
    3: [23, 42, 27, 6],
    4: [13, 22, 14, 13],
    5: [33, 5, 57, 19]
}

# --- Problem parameters ---
n = len(jobs)
m = len(next(iter(jobs.values())))
job_ids = list(jobs.keys())

# --- Compute Big M as: n * max(p_ik) ---
max_p = max(max(times) for times in jobs.values())
M = n * max_p
print(f"Computed Big M = {M}")

# --- Gurobi model ---
model = Model("Manne_PFSP")
model.setParam('OutputFlag', 1)  # Set to 0 to suppress solver output

# Variables
C = model.addVars(job_ids, range(m), name="C", vtype=GRB.CONTINUOUS)
z = model.addVars(itertools.combinations(job_ids, 2), vtype=GRB.BINARY, name="z")
Cmax = model.addVar(name="Cmax", vtype=GRB.CONTINUOUS)

# --- Constraints ---

# 1. Job flow across machines
for i in job_ids:
    for k in range(1, m):
        model.addConstr(C[i, k] >= C[i, k-1] + jobs[i][k], name=f"flow_{i}_{k}")

# 2. Precedence constraints
for i, j in itertools.combinations(job_ids, 2):
    for k in range(m):
        model.addConstr(C[j, k] >= C[i, k] + jobs[i][k] - M * (1 - z[i, j]), name=f"prec_{i}_{j}_{k}")
        model.addConstr(C[i, k] >= C[j, k] + jobs[j][k] - M * z[i, j], name=f"prec_{j}_{i}_{k}")

# 3. Makespan definition
for i in job_ids:
    model.addConstr(Cmax >= C[i, m-1], name=f"Cmax_{i}")

# Objective: minimize makespan
model.setObjective(Cmax, GRB.MINIMIZE)

# Solve
model.optimize()

# --- Output ---
if model.status == GRB.OPTIMAL:
    print(f"\nOptimal Makespan (Cmax): {Cmax.X:.1f}")
    for i in job_ids:
        for k in range(m):
            print(f"Job {i} on Machine {k+1}: Completion = {C[i, k].X:.1f}")
else:
    print("No optimal solution found.")
```
